## Work log (compulsory activities)

The work in this part of the project focused on building a data pipeline from the Elhub API into Cassandra and MongoDB, and then visualising the results in a Streamlit app. I started by reusing the GitHub repository and Streamlit app from the previous project part, and I created a new branch to keep the updates separate until peer review and feedback are finished.

In the Jupyter Notebook, I first familiarised myself with the Elhub API documentation, paying special attention to time encoding, hourly granularity, and the limitations on time periods for each request. I then implemented a function to request hourly production data using the `PRODUCTION_PER_GROUP_MBA_HOUR` endpoint for all price areas for the year 2021. The JSON response was filtered to extract only the `productionPerGroupMbaHour` list, which I converted into a Pandas DataFrame. I made sure to include the relevant columns such as price area, production group, timestamp, and quantity in kWh.

Next, I used Spark to write this DataFrame into a Cassandra table and verified that the Spark–Cassandra connection worked correctly. From Cassandra, I used Spark again to select only the columns `priceArea`, `productionGroup`, `startTime`, and `quantityKwh`. Based on this extracted dataset, I created two visualisations in the notebook: a pie chart showing the total production in 2021 for a chosen price area, broken down by production group, and a line plot showing hourly production for the first month of the year, with separate lines for each production group. These plots were used to verify that the data looked sensible over time and across groups.

After validating the data, I inserted the Spark-extracted dataset into MongoDB, preparing it for use in the Streamlit app. In the app, I updated page four to connect to MongoDB (using secrets instead of hard-coding credentials) and built a layout with two columns. On the left, radio buttons allow the user to select a price area and display a pie chart similar to the one in the notebook. On the right, I implemented pills to select which production groups to include and an additional selection widget for choosing a month, which together control a line plot for the chosen subset of data. Finally, I added an expander at the bottom of the page documenting the data source (Elhub) and briefly describing how the data is processed before visualisation.
