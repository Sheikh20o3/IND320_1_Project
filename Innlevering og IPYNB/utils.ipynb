```python
import os
import pandas as pd
import streamlit as st

    # Cache the function's output to improve performance in Streamlit applications.
@st.cache_data(show_spinner=False)
def load_data():
    # Construct the full file path dynamically. 
    # Read the CSV file into a DataFrame.
    path = os.path.join(os.path.dirname(__file__), "open-meteo-subset.csv")
    df = pd.read_csv(path)

    # Attempt to find a date/time column in the DataFrame.
    date_like = [c for c in df.columns if any(k in c.lower() for k in ["date", "time", "datetime", "timestamp"])]
    if date_like:
    # If a potential date/time column is found, select the first one.
        dcol = date_like[0]
        # Convert the selected column to datetime objects.
        df[dcol] = pd.to_datetime(df[dcol], errors="coerce")
        # Check if any successful datetime conversion occurred.
        if df[dcol].notna().any():
            df = df.sort_values(dcol)
        # Create a new 'month' column by converting the datetime to a monthly period string
            df["month"] = df[dcol].dt.to_period("M").astype(str)
    else:
        df["month"] = "Unknown"

    return df
```
